```python
import sqlite3
import pandas as pd

# Specify the CSV file name (update the path if needed)
csv_file = "/Users/ileene/Library/CloudStorage/OneDrive-UniversitasCiputra/kcg/kcg-llm/retail_data/dataset/Sheet1-Retail_Dataset2.csv"

# Read the CSV into a DataFrame
df = pd.read_csv(csv_file)

# Check the columns to ensure they match your expected schema
print("CSV Columns:", df.columns.tolist())

# Connect to (or create) the SQLite database
conn = sqlite3.connect("retail_sales.db")

# Write the DataFrame to a new table called 'sales'
df.to_sql("sales", conn, if_exists="replace", index=False)

# Optional: Check that the table was created and print the first few rows
print(pd.read_sql_query("SELECT * FROM sales LIMIT 5", conn))

# Close the connection
conn.close()
```
