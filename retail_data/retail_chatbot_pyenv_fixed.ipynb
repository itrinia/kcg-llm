```python
import sqlite3
import pandas as pd

# Load CSV into a database
df = pd.read_csv("Sheet1-1-Retail_Dataset2.csv")
conn = sqlite3.connect("retail_sales.db")
df.to_sql("sales", conn, if_exists="replace", index=False)
```

```python
from langchain_ollama import OllamaLLM
from langchain_community.agent_toolkits.sql.toolkit import SQLDatabaseToolkit
from langchain_community.agent_toolkits.sql.base import create_sql_agent
from langchain_community.utilities import SQLDatabase

llm = OllamaLLM(model="llama3")
db = SQLDatabase.from_uri("sqlite:///retail_sales.db")
toolkit = SQLDatabaseToolkit(db=db, llm=llm)

agent_executor = create_sql_agent(
    llm=llm,
    toolkit=toolkit,
    verbose=True
)

response = agent_executor.run("Which product category had the highest order demand in March 2016?")
print(response)
```
